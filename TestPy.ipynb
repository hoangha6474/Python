# Bài tập Python: Quản lý điểm số học sinh

## Yêu cầu thực hiện

1. **Khai báo biến:**
   - `student_name` (chuỗi): Tên học sinh.
   - `math_score`, `science_score` (số thực): Điểm Toán và Khoa học.

2. **In thông tin ban đầu (2 cách):**
   - **Cách 1:** Dùng dấu `+` để nối chuỗi (cần ép kiểu `str()`).
   - **Cách 2:** Dùng `f-string`.

3. **Tính toán cơ bản:**
   - In ra Tổng (`+`), Hiệu (`-`), Tích (`*`), Thương (`/`) của 2 môn.

4. **Cập nhật dữ liệu:**
   - Cộng thêm 1 điểm cho `math_score` và in lại thông tin.

5. **Đánh giá học lực (if-elif-else):**
   - Tính điểm trung bình mới.
   - Xếp loại:
     - ĐTB $\ge$ 8.0: `"Xuất sắc"`
     - ĐTB $\ge$ 6.5: `"Khá"`
     - ĐTB $\ge$ 5.0: `"Trung bình"`
     - Còn lại: `"Cần cố gắng"`

```python
#Khai bao bien
student_name = "Hoang"
math_score, science_score = 8.5, 7.0
print("Tên:", student_name, "- Điểm Toán:", math_score, "- Điểm Khoa học:", science_score)
print("Tổng điểm", math_score + science_score)
print("Hiệu điểm", math_score - science_score)
print("Tích điểm", math_score * science_score)
print("Thương điểm", math_score // science_score)
math_score += 1
print(math_score)
diemtrungbinh = 8.25
if diemtrungbinh >= 8.0:
    print("Xuat Xac")
elif diemtrungbinh >= 6.5:
    print("Kha")
elif diemtrungbinh >= 5.0:
    print("Trung Binh")
else:
    print("Can co gang")
```
