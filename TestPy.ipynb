# Day3 : Yêu cầu: Viết chương trình Python bao gồm:

- Khai báo các hàm activation function phổ biến: sigmoid, tanh, Softplus, ReLU, LeakyReLU, PReLU, Swish. (Công thức chi tiết xem lại ở phần bài tập)
    
- Thiết kế một hàm tổng quát activate nhận đầu vào là tên activation function (dạng chuỗi) và giá trị x (cùng tham số bổ sung nếu cần), sau đó sử dụng if-elif để gọi hàm tương ứng.

```python
import math

# Sigmoid dùng để ánh xạ một số thực bất kì thành một giá trị nằm trong 
def sigmoid(x):
    return 1/ (1 + math.exp(-x))
# Tanh (tang hyperbol) là một hàm toán học trả về giá trị nằm trong khoảng từ -1 đến 1, có đồ thị hình chữ S đối xứng qua gốc tọa độ
def tanh(x):
    return math.tanh(x)
# Softplus là một hàm kích hoạt (activation function) mượt mà trong học máy và mạng nơ-ron, đóng vai trò là phép xấp xỉ khả vi của hàm ReLU
def softplus(x):
    return math.log(1 + math.exp(x))
# ReLU là hàm kích hoạt phi tuyến phổ biến nhất trong học sâu, được định nghĩa bằng công thức f(x) = max(0,x)
def ReLU(x):
    return max(0.0, float(x))
# LeakyReLU là một biến thể của hàm kích hoạt ReLU trong mạng nơ-ron sâu. Thay vì gán giá trị bằng 0 cho các đầu vào âm như ReLU truyền thống, hàm này cho phép một lượng nhỏ tín hiệu âm đi qua bằng cách nhân chúng với một hệ số dốc nhỏ (thường là 0,01)
# LeakyReLU: f(x) = x nếu x >= 0, ngược lại alpha * x
def LeakyReLU(x, alpha = 0.01):
    return float(x) if x >= 0 else float( alpha * x)
# PReLU là một hàm kích hoạt trong học sâu(deep learning), đóng vai trò là một biến thể nâng cao của hàm ReLU và ReakyReLU truyền thống
# PReLU: Tương tự LeakyReLU nhưng alpha là tham số có thể học được
def PReLU(x, alpha=0.25):
    return float(x) if x >= 0 else float(alpha * x)
# Swish là một hàm kích hoạt (activation function) mượt do các nhà nghiên cứu tại Google Brain đề xuất vào năm 2017
# Swish: f(x) = x * sigmoid(beta * x)
def swish(x, beta=1.0):
    return x * sigmoid(beta * x)
```

### Bài 3: Tính $e^x$ bằng chuỗi Taylor

**Yêu cầu:** Viết hàm `calculate_exp(x, n)` sử dụng vòng lặp `for` để tính gần đúng $e^x$ với $n$ số hạng đầu tiên của chuỗi Taylor.

- Tính $e^2$ với $n = 10$.
- So sánh với giá trị thực từ `math.exp(2)`.

```python
# BAI 1
import math
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
factorial(6)
def calculate_e(n):
    e_approx = 0.0
    for k in range(n):
        e_approx += 1/ factorial(k)
    return e_approx
calculate_e(8)
```

```python
# Bai 3
import math
def calculate_exp(x, n):
    total = 0.0
    term = 1.0 
    for k in range(n):
        if k > 0:
            term *= x / k
        total += term
    return total
```

## Bài tập 1: Tìm tất cả các số nguyên tố trong khoảng từ 1 đến N

#### Mô tả bài toán
Viết một chương trình yêu cầu người dùng nhập một số nguyên dương $N$, sau đó in ra tất cả các số nguyên tố từ $1$ đến $N$. Một số nguyên tố là số chỉ chia hết cho $1$ và chính nó (ví dụ: $2, 3, 5, 7, \dots$).

#### Mô tả Input, Output
* **Input:** Một số nguyên dương $N$ (do người dùng nhập).
* **Output:** Các số nguyên tố từ $1$ đến $N$, cách nhau bởi dấu phẩy và khoảng trắng (`", "`).
* **Ví dụ:**
  * **Input:** `10`
  * **Output:** `2, 3, 5, 7`

---

#### Gợi ý làm bài
Một số $n$ là số nguyên tố nếu $n > 1$ và không tồn tại số nguyên $d$ từ $2$ đến $\sqrt{n}$ sao cho $n$ chia hết cho $d$.

* Với mỗi số $n$ từ $2$ đến $N$, thử chia $n$ cho các số từ $2$ đến $\sqrt{n}$.
* Nếu không có số nào trong khoảng này mà $n$ chia hết (dư bằng $0$), thì $n$ là số nguyên tố.
* Nếu tồn tại ít nhất một số mà $n$ chia hết, thì $n$ không phải số nguyên tố.

## Bài tập 2: Tính tổng các số hoàn hảo nhỏ hơn N

#### Mô tả bài toán
Viết một chương trình yêu cầu người dùng nhập một số nguyên dương $N$, sau đó tính tổng tất cả các số hoàn hảo nhỏ hơn $N$. Một số hoàn hảo là số mà tổng các ước số thực sự của nó (không tính chính nó) bằng chính số đó (ví dụ: $6 = 1 + 2 + 3$).

#### Mô tả Input, Output
* **Input:** Một số nguyên dương $N$ (do người dùng nhập).
* **Output:** Tổng của các số hoàn hảo nhỏ hơn $N$.
* **Ví dụ:**
  * **Input:** `10`
  * **Output:** `6` (vì $6$ là số hoàn hảo duy nhất nhỏ hơn $10$)

---

#### Gợi ý làm bài
Một số $n$ là số hoàn hảo nếu tổng các ước số thực sự của $n$ (từ $1$ đến $n - 1$) bằng $n$.

* Với mỗi số $n$, tìm tất cả các ước số thực sự bằng cách thử chia $n$ cho các số từ $1$ đến $n - 1$.
* Nếu $n$ chia hết cho một số $d$ (dư bằng $0$), thì $d$ là ước số thực sự.
* Tính tổng các ước số này và so sánh với $n$.


```python
# Solution
# Bai 1
from math import *

def find_primes(n):
    if n < 2:
        return False
    for i in range(2, int(sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
N = int(input())
result = []
for k in range(1, N + 1):
    if find_primes(k):
        result.append(str(k))
print(result)
```
